```python
import os
import shutil
import re
import logging
import datetime

root = "D:\\#H\\本子\\"
include = []

logFilename = datetime.datetime.now().strftime("%Y%m%d_%H%M%S_organize.log")
logging.basicConfig(level=logging.INFO,
                    format='%(asctime)s - %(levelname)s - %(message)s',
                    datefmt='%Y-%m-%d %H:%M:%S',
                    filename=logFilename)
logger = logging.getLogger()
logger.setLevel(logging.INFO)

logger.info("Start")

# Get all folders in root path
rootFolders = [os.path.join(root, name) for name in os.listdir(root) if os.path.isdir(os.path.join(root, name))]

for rootFolder in rootFolders:
    logger.info(f"Processing {rootFolder}")

    # Get all files in rootFolder
    files = []
    for dirpath, _, filenames in os.walk(rootFolder):
        for filename in filenames:
            files.append(os.path.join(dirpath, filename))

    # Skip if no files found
    if len(files) < 1:
        logger.info(f"No files detected in {rootFolder}.")
        continue

    for file in files:
        fileName = os.path.basename(file)
        pattern = r"\[([^\[\]\(\uff08]+)((\s)(\(|\uff08)[^\[\]\(\)]+(\)|\uff09))?\]"
        match = re.search(pattern, fileName)
        if match:
            author = match.group(1)

            targetFolder = os.path.join(rootFolder, author)
            targetFile = os.path.join(targetFolder, fileName)

            # Create target folder if not exists
            if not os.path.exists(targetFolder):
                logger.info(f"Create folder: {targetFolder}")
                os.makedirs(targetFolder, exist_ok=True)

            # Check if file already exists at target location
            if os.path.exists(targetFile):
                logger.info(f"File already exists at target location: {targetFile}")
                continue

            logger.info("Move file:")
            logger.info(f"From: {file}")
            logger.info(f"To  : {targetFile}")
            try:
                shutil.move(file, targetFile)
            except Exception as e:
                logger.error(f"Error moving file: {str(e)}")
            logger.info("-------------------------")

    logger.info("Delete empty folders")
    for dirpath, dirnames, _ in os.walk(rootFolder, topdown=False):
        for dirname in dirnames:
            dirToCheck = os.path.join(dirpath, dirname)
            if not os.listdir(dirToCheck):
                logger.info(f"Deleting empty folder: {dirToCheck}")
                os.rmdir(dirToCheck)

logger.info("End")
```
